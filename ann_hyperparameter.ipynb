# 1. select optimizer

```python
# create keras tuner model
from keras.src.metrics import metric
def build_model(hp):
  model = Sequential()

  model.add(Dense(32,activation='relu',input_dim=8))
  model.add(Dense(1,activation='sigmoid'))

  optimizer = hp.Choice('optimizer', values=['adam' , 'sgd', 'rmsprop', 'adadelta'])
  model.compile(optimizer=optimizer , loss = 'binary_crossentropy', metrics  = ['accuracy'])
  return model
```

```python
#create tuner object, input is given as constructor
#                       func_name , objective wht we want to increase or decrease , kitne bar krna hai
tuner = kt.RandomSearch(build_model , objective='val_accuracy', max_trials=5)
```

```python
#search for the best model, with best accuraccy
tuner.search(x_train, y_train, epochs =5,validation_data=(x_test, y_test) )
```

```python
# to get the best hyperparameter
tuner.get_best_hyperparameters()[0].values
```

```python
#to get the model with best hyperparameter
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.summary()
```

```python
model.fit(x_train , y_train , batch_size=32, epochs=100, initial_epoch=6 , validation_data=(x_test,y_test))
```

# **2. no of neurons in each layer**

```python
def build_model(hp):
  model = Sequential()

  units = hp.Int('units' , min_value=8, max_value=128, step=8)  #start no , last no , step

  model.add(Dense(units=units , activation='relu', input_dim=8))
  model.add(Dense(1,activation='sigmoid'))


  model.compile(optimizer='rmsprop', loss='binary_crossentropy',metrics=['accuracy'])
  return model
```

```python
tuner = kt.RandomSearch(build_model,
                        objective='val_accuracy',
                        max_trials=5,
                        directory='mydir',
                        project_name='Deepali')
```

```python
tuner.search(x_train, y_train, epochs=5, validation_data=(x_test, y_test))
```

```python
model.summary()
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.fit(x_train, y_train, batch_size=32, epochs=100, initial_epoch=6, validation_data=(x_test, y_test) )
```

how to select no of layers

```python
def build_model(hp):

  model = Sequential()

  model.add(Dense(104, activation='relu', input_dim=8))

  for i in range(hp.Int('num_layers', min_value=1, max_value=10)):
    model.add(Dense(104, activation='relu'))

  model.add(Dense(1, activation='sigmoid'))

  model.compile(optimizer='rmsprop', loss='binary_crossentropy',metrics=['accuracy'])
  return model
```

```python
tuner = kt.RandomSearch(build_model,
                        objective='val_accuracy',
                        max_trials=5,
                        directory='mydir',
                        project_name='Num_layers')
```

```python
tuner.search(x_train, y_train, epochs=5, validation_data=(x_test, y_test))
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.fit(x_train, y_train, batch_size=32, epochs=100, initial_epoch=6, validation_data=(x_test, y_test) )
```
